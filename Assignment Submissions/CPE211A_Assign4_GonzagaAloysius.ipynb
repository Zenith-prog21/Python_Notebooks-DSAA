### ***Generated Array of Numbers***


```python
numbers = [1,7,0,7,1,8]
```

1. ### **Insert number:**

- The Method `.append` inserts a number at the end of an array. In this example, the method inserts the number 13 at end of the list which is index 6.

```python
numbers.append(13)
print("Insert: ", numbers)
```

2. ### **Insert number at a given position:**

- The Method `.insert()` inserts a number at a given index. In the example below, the method `numbers.insert(2, 7)` inserts the number 7 at index 2, shifting all existing elements at and after that index to the right.

```python
numbers.insert(2, 7)
print("Insert at position: ", numbers)
```

3. ### **Remove last element:**

- The Method `.pop()` removes the number at the last index. In this example below, the method `numbers.pop()` removes the last element with the value 13, off the array.

```python
numbers.pop()
print("Remove last element: ", numbers)
```

4. ### **Remove element at any position:**

- The Method `.pop(index)` could also be used to remove the value from a target index. In this example, the method `numbers.pop(4)` pops or removes the value 7 at index of 4.

```python
numbers.pop(4)
print("Remove element at any position: ", numbers)
```

5. ### **Remove a specified element from the array:**

- The Method '.remove()' is used to remove the first occurence from left to right of a specific element of an array by its value. In the given example below, the method `numbers.remove(7)` removes the first encountered value of 7 when traversing the array from left to right.

```python
numbers.remove(7)
print("Remove a specified element from the array: ", numbers)
```

6. ### **Clears an array:**

- The Method '.clear()' empties an array. In the example below, `numbers.clear()` served its purpose on clearing the array and potentially reusing it in the future of this code.

```python
numbers.clear()
print("Clears an array: ", numbers)
```
