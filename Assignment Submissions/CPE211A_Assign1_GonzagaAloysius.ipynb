1. Shows that parenthesis is solved first before exponential
2. Shows that the exponential is solved first before multiplication
3. Shows that division -- which is the same level of operator precedence as multiplication -- is indeed solved first before the addition

```python
x = 10

print(f"1. Parenthesis:    (1 + 1)^2   = {(2+1)**2}")
print(f"2. Exponential:    x * 2^2     = {x * 3**2}")
print(f"3. Multiplication: x + 1 / 2   = {x + 1 / 2}")
print(f"4. Addition:       x - 3       = {x + 3}")
```

## **M. If Statement**

***Explanation:***

If statements are used when we want the program to make a decision whether a given condition is True or False.

The given condition is placed after the if statement, and it checks the output whether it gives a True or a False, then the lines that are inside the brackets below the if statement will be executed.

If the condition on the if statemnt returns a false, we can add an else statement in order to have choices on what to do if the conditions does not work. However, stacking many else statements could make your code more inefficient, thus we can use elif or else if statements when there are many things to compare to.

else if statments or elif in python, is a function which only checks the condition to whenever the first function before that returns a false or does not satisfy the condition, then it checks on this else but it has another if statement or a condition where it can check if it fits here.

We can also observe that indentation is very important here since Python uses the indentation in order to determine which statements are inside the if, elif or else blocks.

```python
x = int(input('Enter an integer x: ')) # function int(arg) converts the string arg into int type
if x<0:
print('First block: x is negative.')
else:
print('First block: x is non-negative.')
if x<10:
print('Second block: x is between 0 to 9.')
if x<5:
print('Third block: x is between 0 to 4.')
else:
print('Third block: x is between 5 to 9.')
else:
print('Second block: x is greater than 9.')
```

We can see here that this program or lines of codes, satisfies the practices that we've seen so far above. The use of else if here makes the code more efficient and fast whenever one of the if statements aproves a condition, it will not run anything else after that.

```python
grade = int(input('Enter your grade: ')) # function int(arg) converts the string arg into int type
if grade>=95:
print(f'Grade = {grade} is Excellent.')
elif grade>=90 and grade<95:
print('Grade = {} is Very Satisfactory.'.format(grade))
elif grade>=85 and grade<90:
print('Grade = {} is Satisfactory.'.format(grade))
elif grade>=80 and grade<85:
print('Grade = {} is Good.'.format(grade))
elif grade>75 and grade<80:
print('Grade = {} is Fair.'.format(grade))
elif grade==75:
print(f'Grade = {grade} is Passed.')
else:
print(f'Grade = {grade} is failed.')
```
