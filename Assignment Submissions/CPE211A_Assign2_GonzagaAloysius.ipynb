While loops are loops or algorithms that repeat a block of code a certain number of times or indefinitely depending on its use. While loops are crucial in all coding languages, and most of the syntax looks the same, such as in the C language and also the python code below.

In the examples below, it is demonstrated what is the use of while loops and its uses for special cases shown below:

- ***Example 1:*** In the example one, we can use while loops in order to print formatted sentences or phrases, which can benifit the iteration. The iteration works as a stopper with the variable initiated to a certain value `i = 1` and is iterated and increased its number as it is growing or looping through the code, once the case of the while loop `while i<5` becomes false, the loop stops and done looping.
- ***Example 2:*** We can also put the variable and its iteration to use by using the variable iterated to format certain things. In this example, we can create a half pyramid by multiplying the variable iterated to the formatting character `print('*' * i)` , creating a pyramid as the iteration grows bigger `i = i + 1`, thus resulting to a half pyramid or a half tree.

```python
# Example 1.
i = 1
while i<5:
  print("i is: {}".format(i))
  i+=1
```

```python
# R1 Programming Exercise
n = int(input("Enter the number of rows: "))
m = int(input("Enter the number of columns: "))

matrix_1 = []
matrix_2 = []

for row in range(n):
    input_row = []
    for cols in range(m):
        input_row.append(int(input(f"Enter the value for matrix_1 at row {row+1}, column {cols+1}: ")))
    matrix_1.append(input_row)
print("Matrix 1: ")
for row in matrix_1:
    print(row)
print("\n")


for row in range(n):
    input_row = []
    for cols in range(m):
        input_row.append(int(input(f"Enter the value for matrix_2 at row {row+1}, column {cols+1}: ")))
    matrix_2.append(input_row)
print("Matrix 2: ")
for row in matrix_2:
    print(row)
print("\n")

sum_matrix = []

for row in range(n):
    row_sum = []
    for col in range(m):
       row_sum.append(matrix_1[row][col] + matrix_2[row][col])
    sum_matrix.append(row_sum)



print("Sum Matrix: ")
for row in sum_matrix:
    print(row)
```

```python
# S1 Programming Exercise
numbers = [2,2,4,6,3,4,6,5,1]
filtered_list = []

for num in numbers:
    if filtered_list.count(num) == 0:
        filtered_list.append(num)

numbers.sort() 
filtered_list.sort()

print("Original List:", numbers)
print("Filtered List:", filtered_list)
```
