### ***Discussion and Explanation:***

The code below creates the `Stack` class to be used on the following program below. A Stack stores a string or a value into a list using `Last-In, First-out (LiFo)` principle, meaning that the last string or value added onto the stack will be the first one removed.

- `def __init__(self)` - Initializes a list for storage used on the stack
- `def isInStack(self,value)` - Finds a certain value or character inside the stack
- `def isEmpty(self)` - Checks if the stack is empty and returns a boolean value
- `def peek(self)` - Allows the programmer to check the top element without removing it from the stack
- `def size(self)` - Returns the size of the stack
- `def push(self, value)` - Pushes a value onto the top of the stack
- `def pop(self)` - Return and removes the value from the top of the stack

```python
class Stack:
    def __init__(self):
        self.elements = []

    # return True when the value is found in the stack.
    def isInStack(self,value):
        return value in self.elements

    # return True if the stack is empty, False otherwise.
    def isEmpty(self):
        return len(self.elements)==0

    #return the element at the top of the stack without removing it from the stack.
    def peek(self):
        if self.isEmpty():
            return None
        else:
            return self.elements[self.size()-1];

    # return the number of elements in the stack.
    def size(self):
        return len(self.elements)

    # store a new element at the top of the stack
    def push(self, value):
        self.elements.append(value)

    # remove the element at the top of the stack and return it
    def pop(self):
        if self.isEmpty():
            print("Cannot pop. Stack is empty.")
            return None
        else:
            return self.elements.pop()
```

- `def dequeue(self)` - Returns and removes the value from the front of the queue


```python
class Queue:
    def __init__(self):
        self.elements = []

    def isEmpty(self):
        return len(self.elements) == 0

    def size(self):
        return len(self.elements)

    def enqueue(self, value):
            self.elements.append(value)
    
    def dequeue(self):
        if self.isEmpty():
            print("Cannot dequeue. Queue is empty.")
            return None
        else:
            return self.elements.pop(0)
```

Since the stack follows LiFo (Last-in, First-out), the characters are reversed once it is enqueued back to the queue, resulting in a reversed user input value.

            `for char_queue in range(Queue.size()):
                Stack.push(Queue.dequeue())`
    - Step 5: Lastly, we then enqueue the `Stack` back onto the `Queue` with the values returned in reverse order to be printed on the following code
            `for char_stack in range(Stack.size()):
             Queue.enqueue(Stack.pop())`
