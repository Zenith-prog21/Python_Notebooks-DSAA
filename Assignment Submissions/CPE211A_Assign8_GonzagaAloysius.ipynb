#### **Initialization of values**

```python
class Linear_Queue:
    def __init__(self, capacity):
        self.elements = []
        self.size = 0
        self.capacity = capacity
        self.front = 0

    # return True if the stack is empty, False otherwise.
    def isEmpty(self):
        return self.elements==[]  # or   len(self.elements)==0

    # return the number of elements in the stack.
    def size(self):
        return self.size

    # inserts/adds a new element at the REAR of the Queue
    def enqueue(self,new_item):
        if self.capacity == len(self.elements):
            print("Cannot enqueue. Queue is full.")
        else:
            self.elements.append(new_item)
            self.size += 1
            # or  self.elements.insert(0,new_item)

    # removes the element at the FRONT of the Queue and return it
    def dequeue(self):
        if self.isEmpty():
            print("Cannot dequeue. Queue is empty.")
            return None
        else:
            temp = self.elements[self.front]
            self.elements[self.front] = None
            self.front += 1
            self.size -= 1
            return temp  # or   self.elements.pop()
```

```python
class Circular_Queue:
    def __init__(self, capacity):
        self.capacity = capacity
        self.elements = [None] * capacity
        self.front = 0
        self.rear = 0
        self.size = 0
        self.count = 0

    def isInQueue(self, value):
        return value in self.elements

    def isEmpty(self):
        return self.size == 0

    def size(self):
        return self.count

    def enqueue(self, new_item):
        if self.size == self.capacity:
            print("Cannot enqueue. Queue is full.")
            return
        self.elements[self.rear] = new_item
        self.rear = (self.rear + 1) % self.capacity
        self.size += 1
        self.count += 1

    def dequeue(self):
        if self.isEmpty():
            print("Cannot dequeue. Queue is empty.")
            return None
        item = self.elements[self.front]
        self.elements[self.front] = None
        self.front = (self.front + 1) % self.capacity
        self.size -= 1
        self.count -= 1
        return item

    
    
```
