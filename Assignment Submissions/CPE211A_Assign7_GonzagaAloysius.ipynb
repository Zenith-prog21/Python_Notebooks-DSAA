## **Prerequisite Algorithm/Class**

### ***Stack Class***

- The Stack algorithm will be used widely throughout this notebook in order to form and evaluate the corresponding input mathematical expression. Using the advantage LIFO (Last in, First out) principle in stack, the algorithm for infix to postfix and the evaluation takes advantages in the operation of `push()` and `pop()` in order to arrange, and evaluate the corresponding input.

##### ***Operations:***

- `.isInStack()` - Checks whether a value is inside the stack or not, returns a `boolean` value.
- `.isEmpty()`   - Checks if the stack is empty or not, returns a `boolean` value.
- `.peek()`      - Checks the next value that can be popped off from the stack, returns a `boolean` value.
- `.size()`      - Checks the size or the number of elements inside the stack, returns an `Integer` value.
- `.push()`      - Initializes or pushes an input value onto the stack.
- `.pull()`      - Returns and Removes the element at the top of the stack, returns the value of the element.

```python
class Stack:
    def __init__(self):
        self.elements = []
        
    # return True when the value is found in the stack.    
    def isInStack(self,value):
        return value in self.elements
        
    # return True if the stack is empty, False otherwise. 
    def isEmpty(self):
        return len(self.elements)==0
    
    #return the element at the top of the stack without removing it from the stack.
    def peek(self):
        if self.isEmpty():
            return None
        else:
            return self.elements[self.size()-1];
    
    # return the number of elements in the stack.
    def size(self):
        return len(self.elements)
    
    # store a new element at the top of  the stack
    def push(self, value):
        self.elements.append(value)
          
    # remove the element at the top of the stack and return it
    def pop(self):
        if self.isEmpty():
            print("Cannot pop. Stack is empty.")
            return None
        else:       
            return self.elements.pop()
```

## **Infix to Postfix**
### ***Algorithm: Tokenization***
#### ***Procedure:***   

The algorithm below accepts user input and tokenizes or groups it into its corresponding numbers (e.g `90 + 10` is grouped into `['90','+','10']`) into a list. With the numbers grouped, it is easier to arrange things and not confuse numbers greater than a single digit into its own digit. The format will make it easier to arrange the numbers in a postfix form using another algorithm next to the code below.


1. The code accepts user input for the infix mathematical expression and save it into a variable as a string
2. Using a temporary string and a for loop, the code traverses through the contents and groups the numbers together in a list:
    - The loop traverses through the string and inputs the numbers onto the temporary variable `number` until it encounters a non number value
    - Once the code encounters the non number character, it uploads the temporary variable `number`, which has the contents of the bundled number, onto the final list that is going to be used for the next algorithm.
3. The code checks for the last leftover numbers and appends them onto the final output named `grouped_values`

```python
user_input = input("Enter an infix expression: ")
grouped_values = []
parenthesis_count = 0
number = ""

for char in user_input:
    if char.isdigit():
        number += char
    else:
        if len(number) != 0:
            grouped_values.append(number)
            number = ""
        if char.isalpha() or char in "+-*/^":
            grouped_values.append(char)
        if char == '(':
            parenthesis_count += 1
            grouped_values.append(char)
        elif char == ')':
            parenthesis_count -= 1
            grouped_values.append(char)

    if parenthesis_count < 0:
        print("Error: Mismatched parentheses")
        break

if number:
    grouped_values.append(number)


print(grouped_values)


```

### ***Algorithm: Infix to Postfix***

#### ***Procedure:***   

After the operand or numbers, operators, and parenthesis are grouped into a variable named `grouped_values`, we then proceed to the next algorithm which sorts the values and operators of the expression onto its final postfix form. The algorithm below benifits greatly from the tokenization process we did earlier on the code above, this will also make it easier for the next process we will do which is to evaluate the postfix code into its resulting value.


The algorithm will use the output from the code above named `grouped_values` and initialized a directory in order to labelize the precedence of the following operators `+, -, *, /, **` using PEMDAS format `Parenthesis, Exponents, Multiplication & Division, Addition & Subtraction`, but the parenthesis will be approached differently by integrating the process directly onto the code below.

1. Initialized the list `final_output`, the directory `precedence`, and the stack `stack` using the `Stack()` class we initializaed earlier on the code from the top.
2. Using a for loop, we traverse through the tokenized list named `grouped_values` and check if:
- If the parenthesis count is not zero `parenthesis_count != 0` or not balanced, then the expression is immediately invalid, and breaks the loop
- If the value iterated is a number, then it is appended to the list `final_output`
- If the value iterated is a left parenthesis, then the value is pushed into a stack named `stack`
- If the value iterated is a right parenthesis, then the stack is popped and appended into the list `final_output` until it encounters a `(` which the loop stops and the left parenthesis is disregarded
- If the value iterated is an operator;
    - The operator iterated is first checked if it is at higher precedence than the operator that is inside the stack, if true, then it checks if the stack is not empty and if it is not a left parenthesis, else it will be false and the boolean value is stored inside conditions 1 and 2 `condition_1 and condition_2`
    - if the operator iterated is at higher precedance `condition_1 == True`, then it is pushed onto the stack.
    - if the operator iterated is not higher or has equal precedance `condition_2 == True` or is not an exponent, then the operators inside the stack is popped and appended into the `final_output`, and disregrads any left parenthesis inside the stack.
3. After the loop, it is then checked if there are left over operators inside the stack and appends it onto the `final_output`
4. The output is then visualized through a for loop with a print statement.

```python
final_output = []
precedence = {'+': 0, '-': 0, '*': 1, '/': 1, '^': 2}
stack = Stack()

for term in grouped_values:
    if parenthesis_count != 0:
        print("Error: Mismatched Parentheses \nCannot continue the conversion to postfix.")
        break
    if term.isdigit():
        final_output.append(term)

    elif term == '(':
        stack.push(term)

    elif term == ')':
        while stack.peek() != '(':
            final_output.append(stack.pop())
        stack.pop()

    else:
        condition_1 = precedence[term] > precedence[stack.peek()] if not stack.isEmpty() and stack.peek() != '(' else True
        condition_2 = precedence[term] <= precedence[stack.peek()] if not stack.isEmpty() and stack.peek() != '(' else False

        if condition_1:
            stack.push(term)

        elif condition_2:
            while not stack.isEmpty() and stack.peek() != '(' and precedence[term] <= precedence[stack.peek()] and term != '^':
                final_output.append(stack.pop())
            stack.push(term)


while not stack.isEmpty()and stack.peek() != '(':
    final_output.append(stack.pop())


final_output_string = ""
for char in final_output:
    print(char, end=" ")
    final_output_string += char + " "
```

3. The two variables is then solved using the operator from the iterated list `+, -, /, *, **`.
4. The result of the calculation/evaluation of the two variables is then pushed back into the stack for the next evaluation
