## **T. Tuple**

#### ***Discussion and Explanation:***

A tuple in python is a collection of elements, works similarly to a list, where the values inside can be accessed by its corresponding index. However, the main difference between the two is that tuple uses much less memory compared to a list, but its main caveat is that tuples are immutable, meaning that we cannot assign values in specific indecies after the first initialization.

In the examples below, we can see how tuples work and how it differs from a normal list:

- ***Example 1:*** We can initialize a tuple using a parenthesis, instead of a bracket. To initialize a tuple, we do the same syntax as of arrays/list but replace the bracket with a parenthesis `t = (1, 2, 3)`. Accessing the tuple works the same as accessing the indecies of arrays, such as `print(t[0])`

- ***Example 2:*** Similar to an array or list, a tuple can also contain different data types in the same tuple. In the code `t = ('a', True, 5)`, we can see that the tuple is initialized with a string, a boolean, and an integer. Each of the values can still be accessed by its index normally with the same syntax such as `T[0]`, `T[1]`, and `T[2]`, which outputs and prints the values `'a'`, `True`, and `5`.

- ***Example 3:*** The main limitation of a tuple is that it is not mutable. As we can see in the code below, the code tried to initialize a value inside an index of a tuple `t[0] = 'b'`, however, it returns an error because tuples does not allow its values to be changed, hence, immutable.

- ***Example 4:*** Unlike in tuple, lists can initialize different values inside any other index after the initialization of the list, which makes it mutable. In the code below, we can see that in a similar case as example 3, we can successfully initalize values using `t[0] = 'b'` in lists and does not throw an error, which proves the fact that lists are mutable.

- ***Example 5:*** We can also combine tuples with a list. In the example below, the code initialized a list of tuples, each having a value inside them `mypairs = [(1, 2), (3, 4), (5, 6)]`, which can be accessed similarly to 2d arrays using for loops or any other loops using its indecies.

```python
# Example 1:
t = (1, 2, 3)
print(t[0])

# Example 2:
t = ('a', True, 5)
print(t)
print(t[0])
print(t[1])
print(t[2])
```

```python
# Example 3: A tuple is immutable. You cannot modify/change a tuple item.
t = ('a', True, 5)
print(t)
t[0] = 'b'
print(t)
print(t[0])
```

```python
# Example 4: A list is mutable. You can modify/change a list item.
t = ['a', True, 5]
print(t)
t[0] = 'b'
print(t)
print(t[0])

# Example 5: A list of tuples.
mypairs = [(1, 2), (3, 4), (5, 6)]
for item in mypairs:
    print(item)
```

## **U. Unpacking**

#### ***Discussion and Explanation:***

Unpacking in python is a method in separating variables inside an array, or in these cases, separating it from a tuple. Instead of accessing individual indecies using loops and iterations of each indecies, we can assign its values directly to another variable to access it freely and immediatly if needed in the algorithm.


In the examples below, we can see how unpacking works and how it can be used with tuples and loops:

- ***Example 1:*** There are different ways to unpack an array or a tuple. Using the tuple `coordinates = (1, 2, 3)`, the code shows how these different methods on unpacking variables work.
    -  The first one is initializing each of the indecies of the tuple to another variable one by one, as seen in the code `a = coordinates[0]`, `b = coordinates[1]`, and so on. This method works line by line in initializing the variables in order to get or obtain precisely just the needed variables in the tuple.

    - The second one is we can separate it directly using a shortcut code `x, y, z = coordinates`, which separates the variable directly and immediately in corresponding order to the order of variables. Index 0 should be on the variable 'x', index 1 on variable 'y', and index 2 on variable 'z'.

- ***Example 2:*** We can also use unpacking while traversing through a list of tuples. In the code `for (tup1, tup2) in mypairs:`, every tuple inside the list contains two values, which are immediately separated and assigned into `tup1` and `tup2`, successfully unpacking the tuple and list line by line. it is then printed to show that it successfully separated and unpacked the tuple inside the list.

- ***Example 3:*** This example shows a similar process as example 2, but without using the parenthesis around the variables. The code `for tup1, tup2 in mypairs:` still works the same way and unpacks the two values of every tuple into `tup1` and `tup2` sueccessfully. We can also see that the values can be used in any order after being unpacked, since the code prints `tup2` first before `tup1`.

Unpacking can help the code be more efficient and require less indexing when we often need to extract a particular value of the tuple or array.

```python
# Example 1:
coordinates = (1, 2, 3)

a = coordinates[0]
b = coordinates[1]
c = coordinates[2]
print(f'x={a},y={b},z={c}')

x, y, z = coordinates
print(f'x={x},y={y},z={z}')

# Example 2: Iterate through a tuple and UNPACKING a tuple.
mypairs = [(1, 2), (3, 4), (5, 6)]
for (tup1, tup2) in mypairs:
    print(tup1)
    print(tup2)

# Example 3: Tuple UNPACKING.
mypairs = [(1, 2), (3, 4), (5, 6)]
for tup1, tup2 in mypairs:
    print(tup2)
    print(tup1)

```

## **W. Set**

#### ***Discussion and Explanation:***

A set in python is another collection of elements, however its main difference from a list or tuple is that sets only contain one copy of a value, meaning that repeated or duplicate values are not stored multiple times inside the set.

In the examples below, we can see how sets are initialized and how it works.

- ***Example 1:*** An empty set is initialized using `x = set()`. Unlike in list or array where we can normally use `.append()`, a set uses the `.add()` method in order to insert a new value. We can see in the code below that the user uses the `.add()` method to add the values 1-3 in the set.

- ***Example 2:*** Similar to lists and tuples, sets can also contain different data types inside it. In this example, we can see integers, a boolean, a decimal value, and a string all being added into the same set using the `.add()` method.

- ***Example 3:*** We can see here what happens when we add multiple duplicates of a value inside a set. The same values such as `3`, `1`, `4`, `2.5`, `False`, and `'hello'` are added multiple times into the set, however after printing the set, each unique value only appears once, because of its unique property of only keeping one of its copy.

- ***Example 4:*** We can initialize a set using brackets inside the parenthesis of the set `[value1, value2, value3]`. Even through direct initialization of a set with a list of objects, we can see that it still does not accept duplicates inside the set.

- ***Example 5:*** The same technique can be applied with an already initialized list. The list `li` contains multiple repeated values inside, and the code `s = set(li)` converts or initialize the list into a set. After the conversion, all the repeated values are removed, leaving only one copy of every value initialized or in the list itself.

- ### **Return Statement**

    #### ***Discussion and Explanation***


A return statement is used in a function when we want the function to give or return a value back to the code where the function is called. Unlike only printing the value inside the function, returning the value allows it to be stored inside another variable or used again and recycled in another operation.

- ***Example 1:*** The function `square(number)` uses `print(number * number)` in order to directly print the square of the given number. When the function is called inside the print statement `print(square(6))`, the function first prints the result `36`, however because there is no return statement, the function itself has no value to give back, thus the outer print displays `None`.

- ***Example 2:*** Instead of directly printing the result inside the function, the function returns a value by the line `return number * number`. When `square(6)` is called, the result `36` is returned and assigned into the variable `result`. Since the value is now returned instead of only printed, we can store it, print it, or manipulate it for another operation somewhere else in the program.

Return statements are more efficient and useful when we are reusing an operation for different numbers, such as making a calculator or some more specific operations that are repeated several times throughout a code.

```python
# Example 1:
def square(number):
    print(number * number)


print(square(6))

# Example 2:
def square(number):
    return number * number


result = square(6)
print(result)
```

- ### **Lambda Expression**

    #### ***Discussion and Explanation:***


A lambda expression is another method in creating a function in python, but it is mostly used for shorter functions which can be written in only one line as a shortcut. Instead of initializing another full function using `def`, we can directly place the operation inside a lambda expression.

- ***Example 1:*** In the first example, the function `even_bool(num)` is initialized in order to check whether a number is even or not. The expression `num % 2 == 0` checks its modulo or if there is a remainder when dividing it by 2 or not. The function is then passed into `filter()` together with `mylist`, allowing the code to only keep the numbers where `even_bool()` returns true. The resulting list therefore contains the even numbers `[2, 4, 6, 8]`. This demonstrates the standard way of creating a function that is used in the code.

- ***Example 2:*** The second example performs the same operation as example 1, but instead of creating the separate function `even_bool()`, it uses the lambda expression `lambda num: num % 2 == 0` directly inside the function `filter()`. We can see that both examples produce the same result, however the lambda expression makes the code shorter by putting the operation directly inside the function `filter()`.

Lambda expressions are useful in creating a small compact code in order to make it cleaner and efficient for small operations.

```python
# Example 1:
mylist = [1, 2, 3, 4, 5, 6, 7, 8]


def even_bool(num):
    return num % 2 == 0


evens = filter(even_bool, mylist)
print(list(evens))

# Example 2: Using lambda expression
mylist = [1, 2, 3, 4, 5, 6, 7, 8]
evens = filter(lambda num: num % 2 == 0, mylist)
print(list(evens))
```

## **Y. Exception**

#### ***Discussion and Explanation:***

An exception or run-time error is an error that happens while the program is already running. These errors can happen due to invalid inputs or because the program tries to perform an operation that is not possible. In python, we can use a `try-except` block in order to catch these errors and handle them without stopping the whole program.

- ***Example 1:*** In the code below, the program asks the user to input an age using `input()` and converts it into an integer using `int()`. If the user enters a value that cannot be converted into an integer, such as a word or a character, the program results to an error `ValueError` and halts the program to a stop.

- ***Example 2:*** In this example, the same code is placed inside a `try` block, with a `except ValueError` added below or after the block. This makes the program able to catch the `ValueError` whenever the user enters an invalid input. This results to the code not stopping even when the user enters an invalid input, it insteads throw a print `'Invalid age input.'`.

- ***Example 3:*** Another possible exception is shown by adding the code `risk = income / age`. The program already catches a `ValueError`, however if the user enters `0` as the age, the program tries to divide `20000` by zero, which results to another run-time error called `ZeroDivisionError`. Since this error is not yet included in the `except` block, the program still results and halts into an error.

- ***Example 4:*** In order to handle the second exception from example 3, another `except` block is added using `except ZeroDivisionError`. This makes the program able to handle the two different possible errors separately. If the user enters an invalid integer input, the program prints `Invalid age input.`, while if the user enters `0`, the program catches the division by zero error and prints `Age cannot be zero.`. Except blocks can be stacked to handle further errors that could occur on different algorithms.

Exceptions can therefore be handled by using `try-except` blocks, which allows programmers to prepare the program for possible run-time errors and provide another output or action instead of immediately stopping the whole program itself.

```python
# Example 1: EXCEPTION/RUN-TIME ERROR!
age = int(input('Age: '))
print(age)
```

```python
# Example 2: Using try-except block to catch and handle exception.
try:
    age = int(input('Age: '))
    print(age)
except ValueError:
    print('Invalid age input.')
```

```python
# Example 3: ANOTHER EXCEPTION/RUN-TIME ERROR!
try:
    age = int(input('Age: '))
    income = 20000
    risk = income / age
    print(age)
except ValueError:
    print('Invalid age input.')
```

```python
# Example 4: Add a second except block to catch and handle the second exception.
try:
    age = int(input('Age: '))
    income = 20000
    risk = income / age
    print(age)
except ValueError:
    print('Invalid age input.')
except ZeroDivisionError:
    print('Age cannot be zero.')
```

- ### **Inheritance**

    #### ***Discussion and Explanation:***


Inheritance is a feature of classes where another class can obtain or inherit the methods and properties of another class, its parent class. This can be useful so that we do not need to copy and rewrite the same functions repeatedly when creating multiple classes with similar or the same characteristics as the parent class..

- ***Example 1:*** The class `Mammal` contains the common methods `walk()` and `distinguishing_char()`. The Dog and Cat classes are then initialized with inherited functions from the `Mammal` class, which makes them be able to use the functions from the parent class, and also initializing their own functions which are `bark(self)` and `annoying(self)` which are unique to their class but has similarities because they both come from one parent class.

- ***Example 2:*** The class `Person` in the code serves as the parent class of `Student`, `Employee`, and `NonEmployee`, which makes the three classes inherit the functions that exists in the original parent class `Person`. The `Student` and `Employee` class have different versions of the `compute_community_tax(self)` function, which overrides the original function that exists in the parent class. While the `NonEmployee` class only uses `pass` which means that it does not initialize anything of its own and every function that exists in that class is from the parent class `Person`.

Inheritance allows programmers to inherit or use the same template for creating different classes, which can be useful in creating people class who have similar core characteristics, and video games where programmers can use one template for other sprites.

```python
# Example 1:
class Mammal:
    def walk(self):
        print('-mammals walk')

    def distinguishing_char(self):
        print('-has mammary glands')


class Dog(Mammal):
    def bark(self):
        print('-bark at strangers')


class Cat(Mammal):
    def annoying(self):
        print('-annoying to humans')


dog = Dog()
cat = Cat()

print('Dog characteristics:')
dog.walk()
dog.distinguishing_char()
dog.bark()

print('\nCat characteristics:')
cat.walk()
cat.distinguishing_char()
cat.annoying()

# Example 2: INHERITANCE, OVERRIDING
class Person:
    def __init__(self, id, name, addr, phone):
        self.init_common(id, name, addr, phone)

    def init_common(self, id, name, addr, phone):
        self.id = id
        self.name = name
        self.addr = addr
        self.phone = phone

    def compute_community_tax(self):
        print('P 50.00')


class Student(Person):
    def __init__(self, id, name, addr, phone, course, year_level, units):
        self.init_common(id, name, addr, phone)
        self.course = course
        self.year_level = year_level
        self.units = units

    def compute_community_tax(self):
        print('P 5.00')


class Employee(Person):
    def __init__(self, id, name, addr, phone, job_title, rate, hrs_worked):
        self.init_common(id, name, addr, phone)
        self.job_title = job_title
        self.rate = rate
        self.hrs_worked = hrs_worked

    def compute_community_tax(self):
        annual_income = self.rate * 12
        tax = annual_income * 0.1
        print(f'P {tax:.2f}')


class NonEmployee(Person):
    pass


stud = Student('2018-0120', 'Maria Clara', 'Lapasan', '09173334444', 'bscpe', 2, 18)
emp = Employee('2015-0777', 'Juan Cruz', 'Gusa', '09187778888', 'Programmer', 45000, 40)
non_emp = NonEmployee('01234567', 'John Doe', 'Divisoria', '09205551111')

print('Community Tax Computation:')
print('Student: ')
stud.compute_community_tax()
print('Employee: ')
emp.compute_community_tax()
print('Non-Employee: ')
non_emp.compute_community_tax()
```
